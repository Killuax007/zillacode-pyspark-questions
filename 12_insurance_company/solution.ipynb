```python
from pyspark.sql.functions import *

data1 = [
    (1, "Alice", "Smith", 30, "auto"),
    (2, "Bob", "Johnson", 40, "home"),
    (3, "Carol", "Williams", 35, "life")
]

data2 = [
    (4, "Dave", "Brown", 45, "auto"),
    (5, "Eve", "Jones", 55, "health"),
    (6, "Frank", "Davis", 60, "life")
]

columns = [
    "customer_id",
    "first_name",
    "last_name",
    "age",
    "policy_type"
]

input_df1 = spark.createDataFrame(data1, columns)
input_df2 = spark.createDataFrame(data2, columns)


input_df1.union(input_df2).orderBy('age').display()
```
