```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = SparkSession.builder.getOrCreate()

venture_capitalist_data = [
    ("VC1", "VC Firm 1", 1.5),
    ("VC2", "VC Firm 2", 2.0),
    ("VC3", "VC Firm 3", 1.75),
    ("VC4", "VC Firm 4", 2.5),
]

venture_capitalist_df = spark.createDataFrame(
    venture_capitalist_data, ["vc_id", "vc_name", "funding_limit"]
)

funded_startups_data = [
    ("S1", "Startup 1", "VC1", 2.0),
    ("S2", "Startup 2", "VC1", 1.0),
    ("S3", "Startup 3", "VC2", 2.5),
    ("S4", "Startup 4", "VC2", 2.0),
    ("S5", "Startup 5", "VC3", 1.8),
    ("S6", "Startup 6", "VC3", 1.7),
    ("S7", "Startup 7", "VC4", 3.0),
    ("S8", "Startup 8", "VC4", 2.0),
]

funded_startups_df = spark.createDataFrame(
    funded_startups_data, ["startup_id", "startup_name", "vc_id", "funding"]
)

venture_capitalist_df.show()
funded_startups_df.show()
```

```python
avg_funding_df = funded_startups_df.groupBy(col("vc_id")).agg(
    avg("funding").alias("avg_funding")
)
```

```python
venture_capitalist_df.join(avg_funding_df, "vc_id").filter(
    col("avg_funding") > col("funding_limit")
).show()
```
