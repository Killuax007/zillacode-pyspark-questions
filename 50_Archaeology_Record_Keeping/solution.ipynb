```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = SparkSession.builder.getOrCreate()

artifacts = spark.createDataFrame(
    [
        (1, "Pottery", "Prehistoric", "clay", 150),
        (2, "Weapon", "Medieval", "metal", 90),
        (3, "Jewel", "Roman", "gold", 200),
    ],
    ["ID", "Item", "Period", "Material", "Quantity"],
)

artifacts.show()
```

```python
artifacts = artifacts.withColumn("Material", upper(col("Material")))

artifacts = artifacts.filter(col("Quantity") > 100)

artifacts.show()
```
