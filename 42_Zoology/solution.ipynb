```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

# Initialize Spark Session
spark = SparkSession.builder.appName("Zoology").getOrCreate()

# Create AnimalData DataFrame
AnimalData = spark.createDataFrame(
    [
        ("A1", "Lion", 10, 200.5, "Africa"),
        ("A2", "Tiger", 5, 150.3, "Asia"),
        ("A3", "Bear", 7, 180.2, "North America"),
        ("A4", "Lion", 12, 205.7, "Africa"),
        ("A5", "Tiger", 6, 155.1, "Asia"),
    ],
    ["ID", "Species", "Age", "Weight", "Region"],
)

# Create RegionData DataFrame
RegionData = spark.createDataFrame(
    [("Africa", "Hot"), ("Asia", "Temperate"), ("North America", "Cold")],
    ["Region", "Climate"],
)

AnimalData.show()
RegionData.show()
```

```python
Animal_avg = AnimalData.groupBy("Species", "Region").agg(
    avg(col("Age")).alias("AvgAge"),
    avg(col("Weight")).alias("AvgWeight"),
    count(col("Species")).alias("TotalAnimals"),
)
Animal_avg.show()
```

```python
Animal_avg.join(RegionData, "Region").show()
```
