```python

from pyspark.sql.functions import *

user_behavior_data = [
    ("U1", 45, "2023-01-01"),
    ("U1", 60, "2023-01-02"),
    ("U2", 70, "2023-01-03"),
    ("U3", 30, "2023-01-04"),
    ("U2", 50, "2023-01-02")
]

user_behavior_columns = [
    "userId",
    "watchDuration",
    "date"
]

UserBehavior = spark.createDataFrame(
    user_behavior_data,
    user_behavior_columns
)

UserBehavior.groupBy("userId").agg(sum("watchDuration").alias("totalWatchDuration")).display()
```
