```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = (
    SparkSession.builder.master("local[*]").appName("customer_cross_join").getOrCreate()
)

# transactions DataFrame
transactions_data = [
    (1, 500.0, "2023-07-01", 1001),
    (2, 200.0, "2023-07-02", 1002),
    (3, 300.0, "2023-07-03", 1003),
]
transactions = spark.createDataFrame(
    transactions_data, schema=["trans_id", "trans_amt", "date", "cust_id"]
)

# customers DataFrame
customers_data = [
    (1001, "John", "Doe", 30),
    (1002, "Jane", "Smith", 40),
    (1003, "Bob", "Johnson", 50),
]
customers = spark.createDataFrame(
    customers_data, schema=["cust_id", "first_name", "last_name", "age"]
)

transactions.show()
customers.show()
```

```python
transactions.join(customers).show()
```
