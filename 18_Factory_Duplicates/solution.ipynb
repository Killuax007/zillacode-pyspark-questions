```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
# Products DataFrame
products_data = [
    (1, "Widget A", "Type1"),
    (2, "Widget B", "Type1"),
    (3, "Widget C", "Type2"),
    (4, "Widget D", "Type2"),
    (1, "Widget A", "Type1")
]

products_columns = ["ProductID", "ProductName", "Category"]

products_df = spark.createDataFrame(products_data, products_columns)


# Manufacturing Processes DataFrame
manufacturing_processes_data = [
    (1001, 1, "Cutting", 1.5),
    (1002, 2, "Cutting", 1.6),
    (1003, 3, "Cutting", 1.8),
    (1004, 4, "Cutting", 1.5),
    (1005, 1, "Shaping", 2.0)
]

manufacturing_processes_columns = [
    "ProcessID",
    "ProductID",
    "ProcessName",
    "Duration"
]

manufacturing_processes_df = spark.createDataFrame(
    manufacturing_processes_data,
    manufacturing_processes_columns
)

products_df=products_df.dropDuplicates()
products_df.join(manufacturing_processes_df,'ProductID').\
    select('Category','Duration','ProcessID','ProcessName','ProductID' ,'ProductName').display()


```
