```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

spark = SparkSession.builder.appName("Herpetology").getOrCreate()

# Create observations DataFrame
observations = spark.createDataFrame(
    [(1, 100, 1, 55), (2, 101, 2, 35), (3, 100, 1, 45)],
    ["obs_id", "species_id", "location_id", "count"],
)

# Create species DataFrame
species = spark.createDataFrame(
    [(100, "Python"), (101, "Gecko"), (102, "Frog")], ["species_id", "species_name"]
)

observations.show()
species.show()
```

```python
observations.join(species, "species_id").select(
    "count", "location_id", "obs_id", "species_id", "species_name"
).orderBy("count").show()
```
