```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
from pyspark.sql.window import *
```

```python
# Define schema for df_models
models_schema = StructType([
    StructField("Model_ID", StringType(), True),
    StructField("Model_Name", StringType(), True),
    StructField("Model_Type", StringType(), True),
    StructField("Accuracy", DoubleType(), True)
])

# Data for df_models
models_data = [
    ("M1", "ModelA", "Type1", 0.85),
    ("M2", "ModelB", "Type2", 0.78),
    ("M3", "ModelC", "Type1", 0.88),
    ("M4", "ModelD", "Type3", 0.92)
]

# Create DataFrame
df_models = spark.createDataFrame(models_data, schema=models_schema)
df_models.display()
```

```python
# Define schema for df_usage
usage_schema = StructType([
    StructField("Model_ID", StringType(), True),
    StructField("Date", StringType(), True),  # Can be cast to DateType() later if needed
    StructField("Uses", IntegerType(), True)
])

# Data for df_usage
usage_data = [
    ("M1", "2023-01-01", 100),
    ("M1", "2023-01-02", 120),
    ("M2", "2023-01-01", 200),
    ("M3", "2023-01-01", 150),
    ("M4", "2023-01-02", 130)
]

# Create DataFrame
df_usage = spark.createDataFrame(usage_data, schema=usage_schema)
# df_usage.display()

df_mod=df_usage.groupby(col("Model_ID")).agg(sum(col('Uses')).alias('Total_uses'))



df_model_avg_accuracy = df_models.groupBy(
        "Model_Type"
    ).agg(
        avg("Accuracy").alias(
            "Average_Accuracy"
        )
    )

df_models.join(df_mod,'Model_ID','inner').join(df_model_avg_accuracy,'Model_Type','inner')\
    .select('Accuracy','Average_Accuracy','Model_ID','Model_Name','Total_uses').display()
```
