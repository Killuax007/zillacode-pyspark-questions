```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

# create spark session if not already present for this

spark = (
    SparkSession.builder.master("local[*]").appName("space_observatory").getOrCreate()
)

# stars
data_star = [
    (1, "Sun", "Yellow", "Dwarf", 0.0),
    (2, "Sirius", "Blue", "Main Sequence Star", 8.6),
    (3, "Vega", "Blue", "Main Sequence Star", 25.0),
]
cols_star = ["id", "name", "color", "type", "distance"]
df_star = spark.createDataFrame(data_star, schema=cols_star)

# planets
data_planet = [
    (1, "Mercury", 1, "Terrestrial", 0.39),
    (2, "Venus", 1, "Terrestrial", 0.72),
    (3, "Mars", 1, "Terrestrial", 1.52),
    (4, "Sirius B", 2, "Gas Giant", 20.0),
    (5, "Vega b", 3, "Gas Giant", 25.0),
]
cols_planet = ["id", "name", "star_id", "type", "distance"]
df_planet = spark.createDataFrame(data_planet, schema=cols_planet)

# quick check
df_star.show()
df_planet.show()
```

```python
df_result = df_star.join(df_planet, df_star.id == df_planet.star_id).select(
    df_star.name.alias("star_name"),
    df_star.color.alias("star_color"),
    df_star.type.alias("star_type"),
    df_planet.name.alias("planet_name"),
    df_planet.type.alias("planet_type"),
    df_star.distance.alias("distance_star_earth"),
    df_planet.distance.alias("distance_planet_star"),
)

df_result.show()
```
