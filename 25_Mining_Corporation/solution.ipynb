```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

spark = SparkSession.builder.getOrCreate()

mines_data = [
    (1, "Mine Alpha", "Australia"),
    (2, "Mine Beta", "Canada"),
    (3, "Mine Gamma", "South Africa"),
]
mines = spark.createDataFrame(mines_data, schema=["id", "name", "location"])

extraction_data = [
    (1, "2023-06-30", "Gold", 1000.0),
    (2, "2023-06-30", "Silver", 1200.0),
    (3, "2023-06-30", "Diamond", 800.0),
    (1, "2023-06-29", "Gold", 900.0),
    (2, "2023-06-29", "Silver", 1300.0),
    (3, "2023-06-29", "Diamond", 750.0),
]
extraction = spark.createDataFrame(
    extraction_data, schema=["mine_id", "date", "mineral", "quantity"]
)
```

```python
extraction_grp = extraction.groupBy(col("mine_id"), col("mineral")).agg(
    sum(col("quantity")).alias("Total_quantity")
)
```

```python
mines.join(extraction_grp, mines.id == extraction_grp.mine_id, "inner").select(
    "location", "mineral", "Total_quantity"
).show()
```
