```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = SparkSession.builder.appName("CampingSupplies").getOrCreate()

sales_data = [
    ("S1", "P1", "2023-06-01", 10),
    ("S2", "P2", "2023-06-02", 15),
    ("S3", "P3", "2023-06-02", 20),
    ("S4", "P4", "2023-06-01", 12),
    ("S5", "P5", "2023-06-03", 25),
]
df_sales = spark.createDataFrame(
    sales_data, ["sales_id", "product_id", "date", "quantity_sold"]
)

products_data = [
    ("P1", "Camping Tent", "Camping"),
    ("P2", "Hiking Shoes", "Hiking"),
    ("P3", "Fishing Rod", "Fishing"),
    ("P4", "Insulated Bottle", "Hiking"),
    ("P5", "Outdoor Grill", "Camping"),
]
df_products = spark.createDataFrame(
    products_data, ["product_id", "product_name", "product_category"]
)

df_sales.show()
df_products.show()
```

```python
df_joined = df_sales.join(df_products, "product_id", "inner").select(
    "date", "product_category", col("quantity_sold").alias("total_quantity")
)
df_joined.show()
```
