```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import *
from datetime import *
```

```python
mountain_info_data = [
    ("Mount Everest", 8848, "Nepal", "Himalayas"),
    ("Mount Kilimanjaro", 5895, "Tanzania", "Kilimanjaro"),
    ("Mount Denali", 6190, "USA", "Alaska"),
    ("Mount Fuji", 3776, "Japan", "Fuji"),
    ("Mont Blanc", 4808, "France", "Alps"),
]

mountain_info_columns = ["name", "height", "country", "range"]
mountain_climbers_data = [
    ("John",  "Mount Everest",     date(2020, 1, 1), 8.5),
    ("Jane",  "Mount Everest",     date(2022, 2, 2), 9.0),
    ("Jim",   "Mount Kilimanjaro", date(2021, 3, 3), 6.0),
    ("Jess",  "Mount Kilimanjaro", date(2022, 4, 4), 7.0),
    ("Joe",   "Mount Denali",      date(2022, 5, 5), 10.0),
    ("Jill",  "Mount Denali",      date(2021, 6, 6), 11.0),
    ("Jack",  "Mount Fuji",        date(2022, 7, 7), 4.0),
    ("Jules", "Mount Fuji",        date(2021, 8, 8), 5.0),
    ("Jean",  "Mont Blanc",        date(2020, 9, 9), 12.0),
    ("Josh",  "Mont Blanc",        date(2022,10,10), 13.0)
]
mountain_climbers_columns = [
    "climber_name",
    "mountain_name",
    "climb_date",
    "climb_time"
]

mountain_climbers = spark.createDataFrame(
    mountain_climbers_data,
    mountain_climbers_columns
)

mountain_info = spark.createDataFrame(mountain_info_data, mountain_info_columns)
# display(mountain_climbers)
# display(mountain_info)

window_spec = Window.partitionBy("mountain_name").orderBy(desc('climb_date'), desc("climb_time"))
result_df = mountain_climbers.withColumn("rank", rank().over(window_spec)).filter(col('rank')==1)\
    .select("climb_date",'climb_time','climber_name','mountain_name')

display(result_df)
```
