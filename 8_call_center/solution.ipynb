```python

# calls_df
calls_df = spark.createDataFrame([
    (1, 1, "2022-01-01", 100),
    (2, 2, "2022-01-01", 200),
    (3, 1, "2022-01-02", 150),
    (4, 3, "2022-01-02", 300),
    (5, 2, "2022-01-03", 50)
], ["call_id", "cust_id", "date", "duration"])

# customers_df
customers_df = spark.createDataFrame([
    (1, "Alice", "NY", 10, "doctor"),
    (2, "Bob", "CA", 12, "lawyer"),
    (3, "Charlie", "TX", 6, "engineer")
], ["cust_id", "name", "state", "tenure", "occupation"])

joined_cc=calls_df.join(broadcast(customers_df),calls_df.cust_id == customers_df.cust_id)
joined_cc.groupBy('date').agg(count('date').alias('total_nums'),sum('duration').alias('total_duration')).display()
```
