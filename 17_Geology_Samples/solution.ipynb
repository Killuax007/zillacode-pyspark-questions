```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
geo_df = spark.createDataFrame(
    [
        ("S1", "Basalt_450Ma"),
        ("S2", "Sandstone_300Ma"),
        ("S3", "Limestone"),
        ("S4", "Granite_200Ma"),
        ("S5", "Marble_1800Ma"),
    ],
    ["sample_id", "description"],
)
geo_df.withColumn("arr", split(col("description"), "_")).withColumn(
    "age",
    when(size("arr") > 1, regexp_replace(col("arr").getItem(1), "Ma", "")).otherwise(
        col("arr").getItem(0)
    ),
).select('age',when(size("arr") > 1, col('description')).otherwise(
        col("sample_id")
    ).alias("description") , when(size("arr") > 1, col('sample_id')).otherwise(
        lit("")
    ).alias("sample_id")).orderBy("sample_id").display()
```
