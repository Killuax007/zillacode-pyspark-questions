```python
from pyspark.sql.functions import *
```

```python
# Companies DataFrame
companies_data = [
    (1, "AlphaTech", "Technology"),
    (2, "BetaHealth", "Healthcare"),
    (3, "GammaEntertainment", "Entertainment"),
    (4, "DeltaGreen", "Renewable Energy"),
    (5, "EpsilonFinance", "Finance")
]

companies_columns = [
    "company_id",
    "company_name",
    "industry"
]

companies = spark.createDataFrame(
    companies_data,
    companies_columns
)

investments_data = [
    (1, 1, 5000000),
    (2, 2, 3000000),
    (3, 3, 1000000),
    (4, 4, 4000000),
    (5, 5, 2000000)
]

investments_columns = [
    "investment_id",
    "company_id",
    "amount"
]

investments = spark.createDataFrame(
    investments_data,
    investments_columns
)

companies.join(broadcast(investments),'company_id').select("industry",'amount').orderBy('amount').display()
```
