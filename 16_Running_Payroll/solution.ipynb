```python
from pyspark.sql.functions import*
from pyspark.sql.types import *
```

```python
employees_data = [
    (1, "Alice", 25, "Software Engineer"),
    (2, "Bob", 30, "Data Analyst"),
    (3, "Carol", 28, "Product Manager"),
    (4, "Dave", 24, "Software Engineer")
]

employees = spark.createDataFrame(
    employees_data,
    ["employee_id", "name", "age", "position"]
)

payroll_data = [
    (1, 45.0, 30.0),
    (2, 38.0, 25.0),
    (3, 41.5, 35.0),
    (4, 40.0, 28.0)
]

payroll = spark.createDataFrame(
    payroll_data,
    ["employee_id", "hours_worked", "hourly_rate"]
)
```

```python
joined_data=employees.join(payroll,'employee_id')\
    .withColumn("pay",when(col('hours_worked') > 40 ,
                (40*col('hourly_rate'))+
                (col('hours_worked')- 40)*1.5*col('hourly_rate')).otherwise(col('hours_worked')*col('hourly_rate'))).select("employee_id",'name','pay','position')
display(joined_data)
```
