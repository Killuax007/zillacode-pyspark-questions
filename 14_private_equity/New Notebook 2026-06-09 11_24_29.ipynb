```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import *
from datetime import *
```

```python
pe_firms_data = [
    (1, "ABC Fund", 2010, "New York"),
    (2, "XYZ Fund", 2005, "London"),
    (3, "DEF Fund", 2015, "Paris"),
]

pe_firms_columns = ["firm_id", "firm_name", "founded_year", "location"]

pe_firms = spark.createDataFrame(pe_firms_data, pe_firms_columns)

pe_funds_data = [
    (101, 1, "ABC I", 100, 2010, 2015),
    (102, 1, "ABC II", 150, 2015, 2020),
    (103, 2, "XYZ I", 200, 2010, 2018),
]

pe_funds_columns = [
    "fund_id",
    "firm_id",
    "fund_name",
    "fund_size",
    "fund_start_year",
    "fund_end_year",
]

pe_funds = spark.createDataFrame(pe_funds_data, pe_funds_columns)


pe_investments_data = [
    (1001, 101, "Company A", 10, "2012-05-15"),
    (1002, 101, "Company B", 20, "2013-06-20"),
    (1003, 102, "Company C", 30, "2016-07-25"),
]

pe_investments_columns = [
    "investment_id",
    "fund_id",
    "company_name",
    "investment_amount",
    "investment_date",
]

pe_investments = spark.createDataFrame(pe_investments_data, pe_investments_columns)

r1=pe_investments.join(broadcast(pe_funds), "fund_id").orderBy("investment_id")
display(r1)
r1.join(pe_firms,'firm_id','right').display()
```
