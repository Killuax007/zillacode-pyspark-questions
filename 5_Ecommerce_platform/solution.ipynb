```python
products_df = spark.createDataFrame([
    (1, "Apparel", 25.99),
    (2, "Apparel", 35.99),
    (3, "Footwear", 50.00),
    (4, "Footwear", 75.00),
    (5, "Apparel", 20.00)
], ["product_id", "category", "price"])


# orders_df
orders_df = spark.createDataFrame([
    (101, 1, 2),
    (102, 2, 1),
    (103, 1, 3),
    (104, 3, 1),
    (105, 4, 2)
], ["order_id", "product_id", "quantity"])
joined_po=products_df.join(broadcast(orders_df),products_df.product_id == orders_df.product_id)
joined_po.groupBy('category').agg(avg("price").alias('avg_price'),count('category').alias('total_orders')).display()
```
