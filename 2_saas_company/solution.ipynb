```python
customers = spark.createDataFrame([
    (1, "John", "Doe", "john.doe@email.com"),
    (2, "Jane", "Smith", "jane.smith@email.com")
], ["customer_id", "first_name", "last_name", "email"])

# orders
orders = spark.createDataFrame([
    (1001, 1, 101, "2023-01-10"),
    (1002, 2, 102, "2023-01-11")
], ["order_id", "customer_id", "product_id", "order_date"])

# products
products = spark.createDataFrame([
    (101, "Product A", "Category1"),
    (102, "Product B", "Category2")
], ["product_id", "product_name", "category"])

# customers.withColumn('full_name',concat(col('first_name'), lit(' '),col('last_name'))).display()

joined_data=customers.join(broadcast(orders),customers.customer_id==orders.customer_id).join(broadcast(products),orders.product_id==products.product_id)
joined_data.select('category',concat(col('first_name'), lit(' '),col('last_name')).alias('Customer_name'),'email','order_date','order_id','product_name').display()
```
