```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = SparkSession.builder.appName("MortgageInterestRates").getOrCreate()

# Create MortgageDetails dataframe
mortgage_data = [
    ("M1", "Fixed", 4.5),
    ("M2", "Variable", 3.2),
    ("M3", "Adjustable", 2.8),
]
MortgageDetails = spark.createDataFrame(
    mortgage_data, ["MortgageID", "MortgageType", "InterestRate"]
)

# Create UserMortgages dataframe
user_mortgage_data = [("U1", "M1"), ("U2", "M1"), ("U3", "M2"), ("U4", "M3")]
UserMortgages = spark.createDataFrame(user_mortgage_data, ["UserID", "MortgageID"])

MortgageDetails.show()
UserMortgages.show()
```

```python
MortgageDetails.groupBy(
    col("MortgageID"), col("MortgageType").alias("MortgageType")
).agg(sum("InterestRate").alias("Rate_of_intrest")).select(
    "MortgageType",
    "Rate_of_intrest",
).show()
```
