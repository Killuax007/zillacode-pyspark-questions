```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
from datetime import date

data = [
    (1, "The Avengers", "Joss Whedon", date(2022, 5, 6), 1856.45, "Action, Adventure, Sci-Fi"),
    (2, "Black Panther: Wakanda Forever", "Ryan Coogler", date(2022, 11, 10), None, "Action, Adventure, Drama"),
    (3, "Jurassic World: Dominion", "Colin Trevorrow", date(2022, 6, 10), 1234.56, "Action, Adventure, Sci-Fi"),
    (4, "Fantastic Beasts and Where to Find Them 3", "David Yates", date(2022, 11, 4), None, "Adventure, Family, Fantasy"),
    (5, "Sonic the Hedgehog 2", "Jeff Fowler", date(2022, 4, 8), 789.12, "Action, Adventure, Comedy"),
    (6, "The Batman", "Matt Reeves", date(2022, 3, 4), 2345.67, "Action, Adventure, Crime"),
    (7, "Avatar 2", "James Cameron", date(2022, 12, 16), 5678.90, "Action, Adventure, Fantasy, Sci-Fi"),
    (8, "Doctor Strange in the Multiverse of Madness", "Sam Raimi", date(2022, 3, 25), 4567.89, "Action, Adventure, Fantasy")
]

schema = StructType([
    StructField("movie_id", IntegerType(), True),
    StructField("movie_title", StringType(), True),
    StructField("director_name", StringType(), True),
    StructField("release_date", DateType(), True),
    StructField("box_office_collection", FloatType(), True),
    StructField("genre", StringType(), True)
])

movies_df = spark.createDataFrame(data, schema)


null_movies=movies_df.where(col('box_office_collection').isNull())
null_movies.display()
```
