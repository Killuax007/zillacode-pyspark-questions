```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = SparkSession.builder.getOrCreate()
data = [
    (1, "5+3"),
    (2, "hello"),
    (3, "6/3"),
    (4, "world"),
    (5, "2*3+1"),
]

df_math_expr = spark.createDataFrame(data, ["id", "expression"])
df_math_expr.show()
```

```python
valid_expression = col("expression").rlike("^([0-9]+[\+\-\*/])+[0-9]+$")

df_filtered = df_math_expr.filter(valid_expression)

df_filtered.show()
```
