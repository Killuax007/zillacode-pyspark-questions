```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import *

transactions_data = [
    ("CUST1", "PROD1", 2, "2023-01-01"),
    ("CUST1", "PROD2", 1, "2023-01-05"),
    ("CUST2", "PROD3", 5, "2023-02-03"),
    ("CUST3", "PROD1", 4, "2023-02-07"),
    ("CUST1", "PROD3", 3, "2023-02-10")
]

transactions_columns = [
    "customer_id",
    "product_id",
    "quantity",
    "date"
]

transactions_df = spark.createDataFrame(
    transactions_data,
    transactions_columns
)


window_spec = Window.partitionBy("customer_id").orderBy("date")

final_df = transactions_df.withColumn("prev_product", lag("product_id", offset=1, default='None').over(window_spec)) \
             .select("customer_id",'date',concat_ws(lit(" "),col("date"),col('prev_product')).alias('date_and_product'),'prev_product',"product_id",'quantity' )

final_df.display()
```
